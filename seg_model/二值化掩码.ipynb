```python
from pathlib import Path
import numpy as np
from PIL import Image

from mmseg.apis import init_model, inference_model

config = r"Zihao-Configs/ZihaoDataset_DeepLabV3plus_20230818.py"
checkpoint = r"checkpoint/best_mIoU_iter_40000.pth"
device = "cuda:0"  # 或 "cpu"

model = init_model(config, checkpoint, device=device)
```

```python
img_path = r"pred_k31.715_k33.550.jpg"

result = inference_model(model, str(img_path))

pred = result.pred_sem_seg.data.detach().cpu().numpy()
pred = np.squeeze(pred)  # (H,W)

background_id = 0
mask = (pred != background_id).astype(np.uint8) * 255

out_path = r"outputs/testset-pred/mask.png"
Image.fromarray(mask, mode="L").save(out_path)

out_path
```
